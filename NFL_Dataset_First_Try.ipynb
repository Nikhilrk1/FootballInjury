```python
import nflreadpy as nfl
import pandas as pd

rosters = nfl.load_rosters_weekly([2023, 2024]).to_pandas()
print(rosters.shape)
rosters.head()
```

```python
print(list(rosters.columns))
```

```python
stats = nfl.load_player_stats([2023, 2024]).to_pandas()
print(stats.shape)
print(list(stats.columns))
```

```python
snaps = nfl.load_snap_counts([2023, 2024]).to_pandas()
print(snaps.shape)
print(list(snaps.columns))
```

```python
sched = nfl.load_schedules([2023, 2024]).to_pandas()
print(sched.shape)
print(list(sched.columns))
```

```python
base = rosters[["gsis_id", "season", "week", "team", "position"]].drop_duplicates()
print(base.shape)
print(base.duplicated(["gsis_id", "season", "week"]).sum())
base.head()
```

```python
cols = ["gsis_id", "season", "week", "team", "position",
        "birth_date", "height", "weight", "years_exp"]
base = rosters[cols].drop_duplicates()

base["birth_date"] = pd.to_datetime(base["birth_date"], errors="coerce")
season_start = pd.to_datetime(base["season"].astype(str) + "-09-01")
base["age"] = (season_start - base["birth_date"]).dt.days / 365.25

print(base.shape)
print(base.duplicated(["gsis_id", "season", "week"]).sum())
print(base["age"].describe())
```

```python
stat_cols = ["player_id", "season", "week", "carries", "targets", "receptions",
             "rushing_yards", "receiving_yards", "attempts", "sacks_suffered",
             "def_tackles_solo", "def_tackles_with_assist"]

s = stats[stats["season_type"] == "REG"][stat_cols].rename(columns={"player_id": "gsis_id"})
print("dupes in stats:", s.duplicated(["gsis_id", "season", "week"]).sum())

table = base.merge(s, on=["gsis_id", "season", "week"], how="left")
print(table.shape)
print(table[["carries", "targets"]].isna().mean())
```

```python
workload = ["carries", "targets", "receptions", "rushing_yards", "receiving_yards",
            "attempts", "sacks_suffered", "def_tackles_solo", "def_tackles_with_assist"]

table["has_stats"] = table["carries"].notna().astype(int)
table[workload] = table[workload].fillna(0)

print(table["has_stats"].mean())
print(table[workload].describe().loc[["mean", "max"]])
```

```python
players = nfl.load_players().to_pandas()
print(players.shape)
print(list(players.columns))
```

```python
id_map = players[["gsis_id", "pfr_id"]].dropna().drop_duplicates()
print(id_map.shape)
print("repeat pfr_ids:", id_map["pfr_id"].duplicated().sum())

unmatched = ~snaps["pfr_player_id"].isin(id_map["pfr_id"])
print("snap rows with no match:", unmatched.sum(), "of", len(snaps))
```

```python
sn = snaps[snaps["game_type"] == "REG"].merge(
    id_map, left_on="pfr_player_id", right_on="pfr_id", how="inner")

sn = sn[["gsis_id", "season", "week", "offense_snaps", "offense_pct",
         "defense_snaps", "defense_pct", "st_snaps", "st_pct"]]

print("dupes in snaps:", sn.duplicated(["gsis_id", "season", "week"]).sum())
print(sn.shape)

table = table.merge(sn, on=["gsis_id", "season", "week"], how="left")
print(table.shape)
print(table["offense_snaps"].isna().mean())
```

```python
snap_cols = ["offense_snaps", "offense_pct", "defense_snaps", "defense_pct", "st_snaps", "st_pct"]

# check first: among players WITH a snap row, any blanks left?
has_row = table["offense_snaps"].notna()
print(table.loc[has_row, snap_cols].isna().sum())

# then flag and fill
table["has_snaps"] = has_row.astype(int)
table[snap_cols] = table[snap_cols].fillna(0)

print(table["has_snaps"].mean())
print(table[snap_cols].describe().loc[["mean", "max"]])
```

```python
reg = sched[sched["game_type"] == "REG"]

home = reg[["season", "week", "home_team", "roof", "surface", "temp", "wind", "home_rest"]].rename(
    columns={"home_team": "team", "home_rest": "rest_days"})
home["is_home"] = 1

away = reg[["season", "week", "away_team", "roof", "surface", "temp", "wind", "away_rest"]].rename(
    columns={"away_team": "team", "away_rest": "rest_days"})
away["is_home"] = 0

team_games = pd.concat([home, away], ignore_index=True)

print(team_games.shape)
print("dupes:", team_games.duplicated(["team", "season", "week"]).sum())
print(sorted(set(table["team"].dropna()) - set(team_games["team"])))
```

```python
table = table.merge(team_games, on=["team", "season", "week"], how="left")
```

```python
inj = nfl.load_injuries([2023, 2024]).to_pandas()
print(inj.shape)
print(list(inj.columns))
print(inj["game_type"].value_counts(dropna=False))
print(inj["report_status"].value_counts(dropna=False))
```

```python
print(inj["report_primary_injury"].value_counts(dropna=False))
```

```python
inj = inj[inj["game_type"] == "REG"].copy()

print(inj.shape)
print("dupes:", inj.duplicated(["gsis_id", "season", "week"]).sum())
print(inj["practice_status"].value_counts(dropna=False))
print(inj["report_primary_injury"].value_counts().head(15))
```

```python
d = inj[inj.duplicated(["gsis_id", "season", "week"], keep=False)]
print(d[["gsis_id", "season", "week", "team", "report_status", "practice_status", "report_primary_injury"]])

print(pd.crosstab(inj["practice_status"], inj["report_status"].fillna("blank")))
```

```python
# 1. clean junk practice values (stray newline and "Note" become blank)
ps = inj["practice_status"].str.strip()
inj["practice_status"] = ps.mask(ps.isin(["", "Note"]))
rs = inj["report_status"]
inj["report_status"] = rs.mask(rs == "Note")

# 2. collapse duplicates: keep the most severe status per player-week
rank = {"Out": 3, "Doubtful": 2, "Questionable": 1}
inj["sev"] = inj["report_status"].map(rank).fillna(0)
inj = inj.sort_values("sev", kind="stable").drop_duplicates(
    ["gsis_id", "season", "week"], keep="last").drop(columns="sev")

# 3. checks
print(inj.shape)
print("dupes:", inj.duplicated(["gsis_id", "season", "week"]).sum())
print(inj["practice_status"].isna().sum())
non_injury = inj["report_primary_injury"].str.contains("Not injury|Illness|Rest|Personal", case=False, na=False)
print(inj.loc[non_injury, "report_primary_injury"].value_counts())
```

```python
lab = inj[["gsis_id", "season", "week", "report_status", "report_primary_injury"]].copy()

lab["next_listed"] = 1
lab["next_gd_status"] = lab["report_status"].isin(["Questionable", "Doubtful", "Out"]).astype(int)
lab["next_out"] = (lab["report_status"] == "Out").astype(int)
lab["next_non_injury"] = lab["report_primary_injury"].str.contains(
    "Not injury|Illness|Rest|Personal", case=False, na=False).astype(int)

lab = lab[["gsis_id", "season", "week", "next_listed", "next_gd_status", "next_out", "next_non_injury"]]
lab["week"] = lab["week"] - 1

table = table.merge(lab, on=["gsis_id", "season", "week"], how="left")

print(table.shape)
print("lab rows:", len(lab), " matched:", table["next_listed"].notna().sum())
```

```python
chk = lab.merge(table[["gsis_id", "season", "week"]], on=["gsis_id", "season", "week"],
                how="left", indicator=True)
un = chk[chk["_merge"] == "left_only"]

print(len(un))
print((un["week"] + 1).value_counts().sort_index())
```

```python
# 1. remove the label columns from 3o
table = table.drop(columns=["next_listed", "next_gd_status", "next_out", "next_non_injury"])

# 2. each team's next game week
tg = team_games[["team", "season", "week"]].sort_values(["team", "season", "week"]).copy()
tg["next_week"] = tg.groupby(["team", "season"])["week"].shift(-1)
table = table.merge(tg, on=["team", "season", "week"], how="left")

# 3. attach labels using that next game week
lab2 = lab.assign(next_week=lab["week"] + 1).drop(columns="week")
table = table.merge(lab2, on=["gsis_id", "season", "next_week"], how="left")

print(table.shape)
print("matched:", table["next_listed"].notna().sum())
print(table.loc[table["next_week"].isna(), "week"].value_counts())
```

```python
label_cols = ["next_listed", "next_gd_status", "next_out", "next_non_injury"]

model_table = table[table["next_week"].notna()].copy()
model_table[label_cols] = model_table[label_cols].fillna(0).astype(int)

print(model_table.shape)
print(model_table[label_cols].sum())
print(model_table[label_cols].mean())
```

```python
model_table.to_csv("nfl_player_week.csv", index=False)
```

```python
now = inj[["gsis_id", "season", "week", "report_status"]].copy()
now["listed_now"] = 1
now["out_now"] = (now["report_status"] == "Out").astype(int)
now = now[["gsis_id", "season", "week", "listed_now", "out_now"]]

model_table = model_table.merge(now, on=["gsis_id", "season", "week"], how="left")
model_table[["listed_now", "out_now"]] = model_table[["listed_now", "out_now"]].fillna(0).astype(int)

print(model_table.shape)
print(pd.crosstab(model_table["listed_now"], model_table["next_listed"], normalize="index"))
```

```python
new_only = model_table[model_table["listed_now"] == 0].copy()

print(new_only.shape)
print(new_only[label_cols].sum())
print(new_only[label_cols].mean())
```

```python
model_table.to_csv("nfl_player_week.csv", index=False)
new_only.to_csv("nfl_new_injury.csv", index=False)

print(new_only.groupby("has_snaps")["next_listed"].agg(["size", "mean"]))
```

```python
model_table = model_table.sort_values(["gsis_id", "season", "week"]).reset_index(drop=True)

model_table["total_snaps"] = (model_table["offense_snaps"] + model_table["defense_snaps"] + model_table["st_snaps"])

g = model_table.groupby(["gsis_id", "season"])["total_snaps"]
model_table["snaps_last3"] = g.transform(lambda s: s.rolling(3, min_periods=1).sum())

new_only = model_table[model_table["listed_now"] == 0].copy()

print(model_table[["total_snaps", "snaps_last3"]].describe().loc[["mean", "max"]])
print(model_table.loc[model_table["has_snaps"] == 1,
                      ["gsis_id", "season", "week", "total_snaps", "snaps_last3"]].head(8))
```

```python
g2 = model_table.groupby("gsis_id")["listed_now"]
model_table["listings_to_date"] = g2.cumsum()
model_table["listings_last3"] = g2.transform(lambda s: s.rolling(3, min_periods=1).sum())

new_only = model_table[model_table["listed_now"] == 0].copy()

print(model_table[["listings_to_date", "listings_last3"]].describe().loc[["mean", "max"]])
print(new_only.groupby(new_only["listings_to_date"].clip(upper=4))["next_listed"].agg(["size", "mean"]))
```
